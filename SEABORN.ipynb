```python
# Relational plots
# scatter plot
sns.get_dataset_names()
tips=sns.load_dataset("tips")
print(tips.head())

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
plt.show()


#lineplot
flights=sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
plt.show()
```

```python
#BAr plot
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
plt.show()
#box plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"

)
plt.show()
```

```python
# histogram
penguins=sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
print(penguins.head())
plt.show()
```

```python
#Heatmap
flight=sns.load_dataset("flights")
flights_pivot=flight.pivot(index="month",columns="year",values="passengers")
print(flights_pivot)

sns.heatmap(
    flights_pivot,
    cmap="coolwarm",annot=True,fmt="d"
)
plt.title("Heatmap for passengers")
plt.tight_layout()
plt.show()
```

```python
# Recommended approach
fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None,
    ax=ax
)

ax.set_title("Average Total bill by day and sex")
ax.set_xlabel("Day of week")
ax.set_ylabel("Total Bill($)")

ax.legend(title="sex")
plt.tight_layout()
plt.show()
```
