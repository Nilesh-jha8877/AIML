```python
class bank:
  def __init__(self,name, Balance):
    self.name = name
    self.Balance = Balance

  def deposit(self,increment):
    self.Balance = self.Balance + increment
    return self.Balance

  def withdrawl(self,decrement):
    self.Balance = self.Balance + decrement
    return self.Balance

  def display_info(self):
    return {self.name}, {self.Balance}

my_bank = bank("Rahul", 3400)
print(my_bank.withdrawl(10))
print(my_bank.deposit(10))
print(my_bank.display_info())
```

```python
class car:
  def __init__(self, company, model, speed):
    self.company = company
    self.model = model
    self.speed = speed

  def increase_speed(self, increment):
    self.speed = self.speed + increment
    return self.speed

  def decrease_speed(self, decrement):
    if self.speed - decrement < 180:
      self.speed = 180
    else:
      self.speed = self.speed - decrement
    return self.speed


my_car = car("Tata", "Nexon", 200)

print(my_car.model)
print(my_car.speed)
print(my_car.increase_speed(20))
print(my_car.decrease_speed(40))
```

```python
class Person:

  def location (self,loc):
    self.loc = loc
    print(f"Location of person : {self.loc}")

class Student(Person):

  def name (self,name):
    self.name = name

  def school(self,school_name):
    self.school_name = school_name

  def college(self,college_name):
    self.college_name = college_name

obj = Student()
obj.name("Padmavati")
obj.college("Rungta")
obj.loc("Bhilai")
```

```python
class Teacher:
    def teach(self):
        print("Teaching students")
class Researcher:
    def research(self):
        print("Doing research")
class Professor(Teacher, Researcher):
    pass
p = Professor()
p.teach()
p.research()
```

```python
class Fly:
    def fly(self):
        print("Can fly")

class Swim:
    def swim(self):
        print("Can swim")

class Duck(Fly, Swim):
    pass

d = Duck()
d.fly()
d.swim()
```

```python
# Create Grandparent → Parent → Child.
```

```python
class Grandparent:
    def house(self):
        print("Owns a house")

class Parent(Grandparent):
    def car(self):
        print("Owns a car")

class Child(Parent):
    def bike(self):
        print("Owns a bike")

c = Child()
c.house()
c.car()
c.bike()
```

```python
# Create Device → Laptop → GamingLaptop.
```

```python
class device:
  def name(self):
    print("name of device")
class laptop(device):
  def model(self):
    print("name of model")
class gaminglaptop(laptop):
  def select(self):
    print("select the laptop")
g = gaminglaptop()
g.name()
g.model()
g.select()
```

```python
class Device:
    def power_on(self):
        print("Device powered on")

class Laptop(Device):
    def coding(self):
        print("Coding on laptop")

class GamingLaptop(Laptop):
    def play_game(self):
        print("Playing games")

g = GamingLaptop()
g.power_on()
g.coding()
g.play_game()
```
